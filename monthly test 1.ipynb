```python
def is_pal(n):
    return str(n)==str(n)[::-1]
print(is_pal(121))
```

```python
def is_notpal(n):
    return str(n)==str(n)[::-1]
print(is_notpal(321))
```

```python
def print_hollow_square(n):
    for i in range(n):
        for j in range(n):
            if 1==0 or i==n-1 or j==0 or j==n-1:
                print("*",end="")
            else:
                print(" ",end="")
        print()
```

```python
print_hollow_square(5)
```
